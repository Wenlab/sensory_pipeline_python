### check ID using napari

```python
import numpy as np
```

```python
def image_process(image_np, background_value = 110):
    """
    Denoise
    """
    image_np = image_np.astype(np.float32)  # Convert to float
    image_np = np.maximum(0, image_np - background_value)
    return image_np
```

#### 0421_EGCG

```python
# w1
w1_green_path = r"G:\LAB\CODE\sensory_pipeline\ID\image_data\20250421_EGCG\w2\green.npy"
w1_red_path = r"G:\LAB\CODE\sensory_pipeline\ID\image_data\20250421_EGCG\w2\red.npy"
w1_green = np.load(w1_green_path)
w1_red = np.load(w1_red_path)
```

```python
w1_green = image_process(w1_green, background_value=110)
w1_red = image_process(w1_red, background_value=110)
```

```python
w1_red = np.flip(w1_red, axis=3)
```

```python
import napari
viewer = napari.Viewer()
viewer.add_image(w1_green[:1], name='w1_green', colormap='green', blending= 'additive',scale=[1,5,1,1])
viewer.add_image(w1_red[:1], name='w1_red', colormap='red',blending='additive', scale=[1,5,1,1])
```
